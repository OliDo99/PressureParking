```python
import lightgbm as lgb, pandas as pd, numpy as np
```

```python
model = lgb.Booster(model_file="parking_model.txt")

features = ["hour", "day_of_week", "is_weekend", "temperature_c",
            "precipitation_mm", "is_event", "event_capacity"]

total_capacity = 500
```

```python
def ask(timestamp, temperature_c, precipitation_mm, is_event=0, event_capacity=0):
    ts = pd.Timestamp(timestamp)
    row = pd.DataFrame([{
        "hour": ts.hour,
        "day_of_week": ts.dayofweek,
        "is_weekend": int(ts.dayofweek in (5, 6)),
        "temperature_c": temperature_c,
        "precipitation_mm": precipitation_mm,
        "is_event": is_event,
        "event_capacity": event_capacity,
    }])[features]                      
    pred = model.predict(row)[0]
    return float(np.clip(pred, 0, total_capacity))
```

```python
ask("2026-11-06 08:00", temperature_c=14, precipitation_mm=0)
```
